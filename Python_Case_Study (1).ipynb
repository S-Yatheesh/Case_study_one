PYTHON CASE STUDY Q1: BANK ACCOUNT

```python
class BankAccount :
  Accstatus = "active"
  def __init__(self,owner,accno,balance):
    self.owner = owner
    self.accno = accno
    self.balance = balance

  def deposit(self,amount):
    if(amount)>=0:
      self.balance += amount
      print(f"You have deposited {amount:.2f}. Balance: {self.balance:.2f}")
    else:
      print("Please Deposit Positive Values")

  def withdraw(self, amount):
    if(amount) >= 0:
      if self.balance >= amount:

        self.balance -= amount
        print(f"You have withdrewed {amount:.2f}. Balance: {self.balance:.2f}")

      else:
        print("Insufficient balance")
    else:
      print("Please withdraw in positive values")

  def display(self):
    print(f"{self.owner}'s acc no:{self.accno} .The balance is :{self.balance}")

cust1 = BankAccount("Alan",145570,2000)
cust2 = BankAccount("Lena",2145523,2500)
cust3 = BankAccount("Rejn",145590,3800)

cust1.display()
cust1.deposit(400)
cust1.withdraw(590)

cust2.display()
```

PYTHON CASE STUDY Q2: WEATHER STUDY

```python
class Weather :

  def __init__(self,day,temperature,humidity,climate):
    self.date = day
    self.temperature = temperature
    self.humidity = humidity
    self.climate = climate
    self.region = "Trivandrum"

  def analyze_temperature(self,temperature):
    if temperature <20:
      print("cold")
    elif 20<= (temperature) <=25:
      print("pleasant weather")
    elif 25<(temperature)<=30:
      print("warm and humid")
    elif 30<(temperature)<=35:
      print("slightly hot")
    elif 35<(temperature)<=40:
      print("very hot")
    else:
      print("extreme hot")

  def display_humidity(self,humidity):
    if humidity <=30:
      print("dry climate")
    elif 30< humidity <=50:
      print("not much dry")
    elif 50< humidity <= 60:
      print("humid")
    elif 60 < humidity <= 70:
      print("Extreme humid")
    else:
      print("Humidity out of range")
  @staticmethod
  def calculate_averages(weather_list):
        total_temp = sum(obj.temperature for obj in weather_list)
        total_hum = sum(obj.humidity for obj in weather_list)
        count = len(weather_list)

        print(f"\n--- Weekly Average Report ---")
        print(f"Average Temperature: {total_temp/count:.2f}°C")
        print(f"Average Humidity: {total_hum/count:.2f}%")

  def display_report(self):
    print(f"Report for {self.date} in {self.region}:")
    self.analyze_temperature(self.temperature)
    self.display_humidity(self.humidity)

day1 = Weather("2026-06-04", 32, 65, "Sunny")
day2 = Weather("2026-06-05", 40, 48, "Sunny")
weekly_data = [day1, day2]
Weather.calculate_averages(weekly_data)
```
